## What is Matplotlib?

**Matplotlib** is a Python **Library** used for **Visualization** — turning data into charts and graphs.

> 🧠 *A table of 100 numbers is hard to understand at a glance. A single bar chart can show the same information instantly.*

Matplotlib works hand-in-hand with NumPy and Pandas — you'll usually plot NumPy arrays or Pandas columns directly.

## 🔧 Setup


```python

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

%matplotlib inline

print("Setup complete! ✅")
```


---
## 1️⃣ Your First Plot

Every chart starts the same way: give it some data, tell it what kind of chart to draw, then show it.


```python

students = ["Ali", "Sara", "Zain", "Mina", "Bilal"]
marks = [85, 90, 76, 60, 95]

plt.plot(students, marks)
plt.show()
```


That's it — that's a chart! But it's plain. The rest of this notebook is about making charts like this clear, labeled, and easy to read.



---
## 2️⃣ Line Charts

A **line chart** connects data points with a line — best for showing a **trend over time or order** (e.g. marks across the year, temperature over a week).


```python

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
average_marks = [70, 74, 78, 80, 85, 88]

plt.figure(figsize=(8, 4))
plt.plot(months, average_marks, marker="o", color="steelblue", linewidth=2)
plt.title("Average Class Marks Over 6 Months")
plt.xlabel("Month")
plt.ylabel("Average Marks")
plt.grid(True, alpha=0.3)
plt.show()
```

```python

# Plotting TWO lines on the same chart, to compare two groups
group_a = [65, 70, 74, 78, 82, 85]
group_b = [72, 75, 77, 79, 83, 90]

plt.figure(figsize=(8, 4))
plt.plot(months, group_a, marker="o", label="Group A")
plt.plot(months, group_b, marker="s", label="Group B")
plt.title("Group A vs Group B — Average Marks")
plt.xlabel("Month")
plt.ylabel("Average Marks")
plt.legend()        # shows which line is which
plt.grid(True, alpha=0.3)
plt.show()
```


---
## 3️⃣ Bar Charts

A **bar chart** uses bars to compare values across different **categories** (e.g. marks per student, sales per city).


```python

students = ["Ali", "Sara", "Zain", "Mina", "Bilal"]
marks = [85, 90, 76, 60, 95]

plt.figure(figsize=(8, 4))
plt.bar(students, marks, color="mediumseagreen")
plt.title("Marks per Student")
plt.xlabel("Student")
plt.ylabel("Marks")
plt.show()
```

```python

# Horizontal bar chart — useful when category names are long
cities = ["Lahore", "Karachi", "Islamabad", "Multan"]
avg_marks = [82, 85, 78, 74]

plt.figure(figsize=(8, 4))
plt.barh(cities, avg_marks, color="coral")
plt.title("Average Marks by City")
plt.xlabel("Average Marks")
plt.show()
```


---
## 4️⃣ Histograms

A **histogram** shows how numeric values are **distributed** — how many fall into each range.

> 🧠 *Different from a bar chart: a bar chart compares separate categories, a histogram groups numbers into ranges (bins) to show their spread.*


```python

np.random.seed(1)
all_marks = np.random.randint(50, 100, size=200)   # 200 simulated student marks

plt.figure(figsize=(8, 4))
plt.hist(all_marks, bins=10, color="orange", edgecolor="black")
plt.title("Distribution of Marks (200 students)")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.show()
```


---
## 5️⃣ Scatter Plots

A **scatter plot** shows the relationship (**Correlation**) between two numeric variables — each point is one record.

> 🧠 *Does studying more hours relate to higher marks? A scatter plot shows the pattern at a glance.*


```python

np.random.seed(2)
study_hours = np.random.uniform(1, 10, 40)
marks = study_hours * 7 + np.random.normal(0, 8, 40)   # marks roughly follow study hours, with some randomness

plt.figure(figsize=(8, 5))
plt.scatter(study_hours, marks, color="purple", alpha=0.7)
plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours per Week")
plt.ylabel("Marks")
plt.grid(True, alpha=0.3)
plt.show()
```


---
## 6️⃣ Pie Charts

A **pie chart** shows how a whole is divided into parts (percentages/proportions).

> ⚠️ Use pie charts sparingly — they work best with just a few categories.


```python

subjects = ["Math", "Science", "Art"]
student_counts = [3, 3, 2]

plt.figure(figsize=(6, 6))
plt.pie(student_counts, labels=subjects, autopct="%1.0f%%",
        colors=["steelblue", "mediumseagreen", "coral"])
plt.title("Students per Subject")
plt.show()
```


---
## 7️⃣ Customizing Charts

Every chart can be styled: titles, axis labels, colors, legends, and figure size all make a chart easier to read.


```python

students = ["Ali", "Sara", "Zain", "Mina", "Bilal"]
marks = [85, 90, 76, 60, 95]
colors = ["green" if m >= 75 else "red" for m in marks]   # green = pass, red = fail

plt.figure(figsize=(9, 5))                    # figure size in inches
plt.bar(students, marks, color=colors)
plt.title("Pass (Green) vs Fail (Red) — Marks per Student", fontsize=14, fontweight="bold")
plt.xlabel("Student", fontsize=12)
plt.ylabel("Marks", fontsize=12)
plt.axhline(y=75, color="black", linestyle="--", label="Pass Mark (75)")
plt.legend()
plt.ylim(0, 100)
plt.show()
```


---
## 8️⃣ Multiple Charts in One Figure (Subplots)

You can show several charts side by side using **subplots** — useful for comparing different views of the same data at once.


```python

fig, axes = plt.subplots(1, 2, figsize=(12, 4))   # 1 row, 2 columns of charts

# Left chart: bar chart
axes[0].bar(students, marks, color="steelblue")
axes[0].set_title("Marks per Student")
axes[0].set_xlabel("Student")
axes[0].set_ylabel("Marks")

# Right chart: histogram
axes[1].hist(all_marks, bins=10, color="orange", edgecolor="black")
axes[1].set_title("Distribution of All Marks")
axes[1].set_xlabel("Marks")
axes[1].set_ylabel("Frequency")

plt.tight_layout()
plt.show()
```


---
## 9️⃣ Plotting Directly From Pandas

Pandas has a built-in `.plot()` method, so you often don't even need to call Matplotlib functions directly — Pandas does it for you using Matplotlib behind the scenes.


```python

df = pd.DataFrame({
    "Name":  ["Ali", "Sara", "Zain", "Mina", "Bilal", "Hina", "Omar", "Noor"],
    "City":  ["Lahore", "Karachi", "Lahore", "Islamabad", "Karachi", "Lahore", "Multan", "Karachi"],
    "Marks": [85, 90, 76, 60, 95, 88, 72, 81]
})

avg_by_city = df.groupby("City")["Marks"].mean()

avg_by_city.plot(kind="bar", color="mediumseagreen", figsize=(7, 4), title="Average Marks by City")
plt.ylabel("Average Marks")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python

# A Pandas Series/DataFrame can also make a quick line plot directly
df["Marks"].plot(kind="line", marker="o", figsize=(7, 4), title="Marks Across Students (in row order)")
plt.ylabel("Marks")
plt.xlabel("Student Row Number")
plt.grid(True, alpha=0.3)
plt.show()
```


---
## 🔟 Saving a Chart as an Image File

Once you're happy with a chart, you can save it as an image (PNG, JPG, etc.) to use in reports, slides, or a GitHub README.


```python

plt.figure(figsize=(8, 4))
plt.bar(students, marks, color="steelblue")
plt.title("Marks per Student")
plt.xlabel("Student")
plt.ylabel("Marks")
plt.tight_layout()

plt.savefig("marks_chart.png", dpi=150)   # saves to a file in the current folder
print("Chart saved as marks_chart.png ✅")

plt.show()
```

## 📖 Quick-Reference Glossary Recap

| Term | Simple Definition |
|---|---|
| Bar Chart | Chart using bars to compare values across categories |
| Correlation | How two variables change together |
| Figure | The overall chart window/canvas |
| Histogram | Chart showing how numeric values are distributed into ranges |
| Legend | A key that explains what each color/line represents |
| Line Chart | Chart connecting points with a line, showing a trend |
| Matplotlib | Library for turning data into charts and graphs |
| Pie Chart | Chart showing parts of a whole as slices |
| Scatter Plot | Chart showing the relationship between two numeric variables |
| Subplot | Multiple charts shown together in one figure |
| Visualization | Turning data into charts/graphs so it's easier to understand |

## ✏️ Practice Exercises

Try these yourself before checking the solutions below.

1. Make a bar chart of the marks `[55, 70, 88, 92, 60]` for students `["Sana", "Ahmed", "Fatima", "Usman", "Iqra"]`.
2. Make a line chart showing temperature over a week: `[20, 22, 19, 23, 25, 24, 21]` for days `["Mon","Tue","Wed","Thu","Fri","Sat","Sun"]`.
3. Make a histogram of 100 random numbers between 1 and 50.
4. Make two subplots side by side: a bar chart and a pie chart, using the same small dataset.


```python

# 1. Bar chart
names = ["Sana", "Ahmed", "Fatima", "Usman", "Iqra"]
ex_marks = [55, 70, 88, 92, 60]

plt.figure(figsize=(7, 4))
plt.bar(names, ex_marks, color="teal")
plt.title("Exercise 1 — Marks per Student")
plt.xlabel("Student")
plt.ylabel("Marks")
plt.show()
```

```python

# 2. Line chart of temperature over a week
days = ["Mon","Tue","Wed","Thu","Fri","Sat","Sun"]
temps = [20, 22, 19, 23, 25, 24, 21]

plt.figure(figsize=(7, 4))
plt.plot(days, temps, marker="o", color="crimson")
plt.title("Exercise 2 — Temperature Over a Week")
plt.xlabel("Day")
plt.ylabel("Temperature (°C)")
plt.grid(True, alpha=0.3)
plt.show()
```

```python

# 3. Histogram of 100 random numbers
np.random.seed(3)
random_numbers = np.random.randint(1, 50, size=100)

plt.figure(figsize=(7, 4))
plt.hist(random_numbers, bins=10, color="slateblue", edgecolor="black")
plt.title("Exercise 3 — Distribution of 100 Random Numbers")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

```python

# 4. Two subplots: bar chart + pie chart
labels = ["Math", "Science", "Art"]
values = [4, 5, 3]

fig, axes = plt.subplots(1, 2, figsize=(11, 4))

axes[0].bar(labels, values, color="goldenrod")
axes[0].set_title("Students per Subject (Bar)")
axes[0].set_ylabel("Count")

axes[1].pie(values, labels=labels, autopct="%1.0f%%", colors=["goldenrod", "seagreen", "indianred"])
axes[1].set_title("Students per Subject (Pie)")

plt.tight_layout()
plt.show()
```
