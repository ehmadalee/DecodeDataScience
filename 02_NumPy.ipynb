## What is NumPy?

**NumPy** (Numerical Python) is a Python **Library** (pre-written, reusable code) built for fast math on numbers and lists.

> 🧠 *Think of a normal Python list as a notebook where you calculate one number at a time. NumPy is like a calculator that does thousands of calculations instantly, all at once.*

### Why not just use normal Python lists?
- NumPy arrays are **much faster** for math
- NumPy lets you do math on a whole array at once (**Vectorization**)
- NumPy powers almost every other data science library (Pandas, Matplotlib, scikit-learn all rely on it)

## 🔧 Setup


```python

import numpy as np
import time

print("NumPy version:", np.__version__)
print("Setup complete! ✅")
```

## 1️⃣ Creating Arrays

An **Array** is a list of items (usually numbers), all the same type, stored together.

> 🧠 *A row of lockers numbered 1–10, each holding one item — that's like an array.*


```python

# Creating an array from a normal Python list
scores = np.array([85, 90, 76, 60, 95])
print("Array:", scores)
print("Type:", type(scores))
```

```python

# A 2D array (like a mini table): rows = students, columns = 3 test scores each
tests = np.array([
    [85, 90, 88],
    [76, 80, 78],
    [60, 65, 70]
])
print("2D array (3 students, 3 test scores each):")
print(tests)
```


---
## 2️⃣ Array Attributes

- **Dimension** = how many directions the data extends in (1D = list, 2D = table, 3D = cube)
- **Shape** = the size of data in terms of rows and columns
- **Data Type** = the kind of value stored (int, float, etc.)


```python

print("Shape of 'tests' array:", tests.shape)        # (rows, columns)
print("Number of dimensions:", tests.ndim)            # 2D
print("Total number of elements:", tests.size)        # 9 elements
print("Data type of elements:", tests.dtype)           # int64 or similar
```


---
## 3️⃣ Array Creation Shortcuts

NumPy gives you quick ways to build arrays without typing every number yourself.


```python

zeros = np.zeros((2, 3))          # 2 rows, 3 columns, all zeros
ones = np.ones((2, 3))            # 2 rows, 3 columns, all ones
identity = np.eye(3)              # 3x3 identity matrix (1s on diagonal)

print("Zeros array:\n", zeros)
print("\nOnes array:\n", ones)
print("\nIdentity matrix:\n", identity)
```

```python

# arange: like Python's range(), but returns a NumPy array
sequence = np.arange(0, 20, 5)    # start=0, stop=20, step=5
print("arange(0, 20, 5):", sequence)

# linspace: evenly spaced numbers between a start and stop value
evenly_spaced = np.linspace(0, 1, 5)   # 5 numbers evenly spaced between 0 and 1
print("linspace(0, 1, 5):", evenly_spaced)
```

```python

# Random numbers (useful for practice data, simulations, testing)
np.random.seed(42)   # makes the "random" numbers repeatable every time we run this
random_scores = np.random.randint(50, 100, size=6)   # 6 random ints between 50 and 100
print("Random scores:", random_scores)
```


---
## 4️⃣ Indexing & Slicing

- **Index** = the position/label used to identify an item
- **Slice** = taking a smaller piece out of an array

> 🧠 *Picking only the first 3 students from a class list of 30.*


```python

scores = np.array([85, 90, 76, 60, 95, 88, 72])

print("Full array:", scores)
print("First item (index 0):", scores[0])
print("Last item (index -1):", scores[-1])
print("Slice [1:4]  (items at index 1,2,3):", scores[1:4])
print("Every second item [::2]:", scores[::2])
```

```python

# Indexing a 2D array: [row, column]
print("2D array:\n", tests)
print("\nStudent 0, all tests:", tests[0])
print("Student 1, test 2 only:", tests[1, 2])
print("All students, test 1 only (column):", tests[:, 1])
```


---
## 5️⃣ Reshaping Arrays

Reshaping changes how the same data is arranged (e.g. turning a flat list into a table), without changing the values themselves.


```python

flat = np.arange(1, 13)   # numbers 1 to 12
print("Flat array:", flat)

reshaped = flat.reshape(3, 4)   # 3 rows, 4 columns
print("\nReshaped into 3 rows x 4 columns:")
print(reshaped)
```

```python

# Transpose: flipping rows into columns and vice versa
print("Original:\n", reshaped)
print("\nTransposed (rows <-> columns):")
print(reshaped.T)
```


---
## 6️⃣ Arithmetic Operations & Broadcasting

- **Vectorization** = doing math on the entire array at once instead of looping item by item
- **Broadcasting** = a NumPy trick that lets you do math on arrays of *different sizes* automatically

> 🧠 *If every student gets 5 bonus marks, broadcasting adds it to everyone instantly — no loop needed.*


```python

scores = np.array([85, 90, 76, 60, 95])

print("Original scores:", scores)
print("Add 5 bonus marks (vectorized):", scores + 5)
print("Scale by 10% curve (broadcasting):", scores * 1.1)
print("Scores squared:", scores ** 2)
```

```python

# Two arrays of the SAME size: element-wise math
test1 = np.array([80, 70, 90])
test2 = np.array([85, 75, 95])

print("Test 1:", test1)
print("Test 2:", test2)
print("Average of both tests (element-wise):", (test1 + test2) / 2)
```

```python

# Broadcasting with a 2D array: add a different bonus to each column
tests = np.array([
    [85, 90, 88],
    [76, 80, 78],
    [60, 65, 70]
])
bonus_per_test = np.array([2, 5, 1])   # different bonus for test 1, 2, 3

print("Original tests:\n", tests)
print("\nBonus per test:", bonus_per_test)
print("\nTests after adding bonus (broadcasting a row across all students):")
print(tests + bonus_per_test)
```


---
## 7️⃣ Aggregation Functions (Statistics)

| Term | Definition |
|---|---|
| **Mean** | The average of a set of numbers |
| **Median** | The middle value when sorted |
| **Standard Deviation** | How spread out numbers are from the average |


```python

scores = np.array([85, 90, 76, 60, 95])

print("Sum:", np.sum(scores))
print("Mean:", np.mean(scores))
print("Median:", np.median(scores))
print("Minimum:", np.min(scores))
print("Maximum:", np.max(scores))
print("Standard Deviation:", round(np.std(scores), 2))
```


---
## 8️⃣ Axis-Based Operations

**Axis** = the direction along which you calculate: rows or columns.

- `axis=0` → down each **column**
- `axis=1` → across each **row**

> 🧠 *Adding up marks "across" each student (row-wise, axis=1) vs "down" each subject (column-wise, axis=0).*


```python

tests = np.array([
    [85, 90, 88],   # student 0
    [76, 80, 78],   # student 1
    [60, 65, 70]    # student 2
])
print("Tests:\n", tests)

print("\nAverage PER STUDENT (across each row, axis=1):", tests.mean(axis=1))
print("Average PER TEST (down each column, axis=0):", tests.mean(axis=0))
```


---
## 9️⃣ Sorting Arrays


```python

scores = np.array([85, 90, 76, 60, 95])

print("Original:", scores)
print("Sorted (ascending):", np.sort(scores))
print("Sorted (descending):", np.sort(scores)[::-1])
```


---
## 🔟 Boolean Masking & Filtering

- **Boolean** = a value that's either `True` or `False`
- **Filter** = picking out only the items that match a certain condition

> 🧠 *Asking "which students scored above 80?" and only keeping those.*


```python

scores = np.array([85, 90, 76, 60, 95, 88, 72])

above_80 = scores > 80          # Boolean mask: True/False for each item
print("Boolean mask (scores > 80):", above_80)

filtered_scores = scores[above_80]   # Filter: keep only the True ones
print("Filtered scores (only above 80):", filtered_scores)
```

```python

# You can also do it in one line
top_scores = scores[scores >= 85]
print("Scores 85 and above:", top_scores)
```


---
## 1️⃣1️⃣ Combining Arrays

- **Concatenate** = joining arrays end-to-end
- **Stack** = joining arrays as new rows or columns


```python

group_a = np.array([85, 90, 76])
group_b = np.array([60, 95, 88])

combined = np.concatenate([group_a, group_b])
print("Concatenated (joined end-to-end):", combined)

stacked_rows = np.vstack([group_a, group_b])     # stack as rows
print("\nStacked as rows:\n", stacked_rows)

stacked_cols = np.hstack([group_a, group_b])     # stack side-by-side
print("\nStacked side-by-side (hstack):", stacked_cols)
```


---
## 1️⃣2️⃣ Vectorization vs. Loops — A Speed Test

Let's actually prove why NumPy is faster. We'll add 1 to every number in a list of 1,000,000 numbers two ways:

1. Using a normal Python loop
2. Using NumPy vectorization


```python

big_list = list(range(1_000_000))
big_array = np.arange(1_000_000)

# Method 1: normal Python loop
start = time.time()
result_loop = [x + 1 for x in big_list]
loop_time = time.time() - start

# Method 2: NumPy vectorization
start = time.time()
result_vectorized = big_array + 1
vector_time = time.time() - start

print(f"Python loop time:       {loop_time:.5f} seconds")
print(f"NumPy vectorized time:  {vector_time:.5f} seconds")
print(f"\nNumPy was about {loop_time / vector_time:.0f}x faster! 🚀")
```
