## What is Pandas?

**Pandas** is a Python **Library** built specifically for working with tables of data.

> 🧠 *Think of Pandas as an Excel spreadsheet you can control entirely with code — sort it, filter it, calculate with it, clean it, all in a few lines.*

Pandas is built on top of **NumPy**, so a lot of the array ideas from `02_NumPy.ipynb` (indexing, slicing, vectorization) carry over directly.

## 🔧 Setup


```python

import pandas as pd
import numpy as np

print("Pandas version:", pd.__version__)
print("Setup complete! ✅")
```


---
## 1️⃣ Series vs DataFrame

| Term | Definition | Real-life example |
|---|---|---|
| **Series** | A single column of data with labels | Just the "Marks" column on its own |
| **DataFrame** | A full table with rows and columns | An entire class record sheet |


```python

# A Pandas Series: a single labeled column
marks_series = pd.Series([85, 90, 76, 60, 95], name="Marks")
print("A Pandas Series:")
print(marks_series)
print("\nType:", type(marks_series))
```

```python

# A Series with custom labels (Index) instead of default 0,1,2...
custom_series = pd.Series([85, 90, 76], index=["Ali", "Sara", "Zain"], name="Marks")
print("Series with custom Index (student names):")
print(custom_series)
```


---
## 2️⃣ Creating a DataFrame

A **DataFrame** is Pandas' main structure — a table made of rows and columns.

We'll build one **Dataset** (a collection of related data) about students, and use it throughout this notebook.

- **Data** = the raw facts/numbers themselves
- **Feature** = a single measurable column (e.g. "Age", "Marks")
- **Column** = a vertical set of values
- **Row** = a horizontal set of values — one full record


```python

data = {
    "Name":   ["Ali", "Sara", "Zain", "Mina", "Bilal", "Hina", "Omar", "Noor"],
    "Age":    [20, 21, 19, 22, 20, 23, 21, np.nan],       # np.nan = missing value
    "City":   ["Lahore", "Karachi", "Lahore", "Islamabad", "Karachi", "Lahore", "Multan", "Karachi"],
    "Marks":  [85, 90, 76, 60, 95, 88, 72, 81],
    "Subject":["Math", "Science", "Math", "Art", "Science", "Math", "Art", "Science"]
}

df = pd.DataFrame(data)
df
```


---
## 3️⃣ Exploring a Dataset

Always look at your data before working with it.

- **Head** = viewing just the first few rows
- **Shape** = size of data in (rows, columns)
- **Index** = the row labels
- **Data Type** = the kind of value each column holds


```python

print("First 3 rows (head):")
display(df.head(3))

print("\nLast 2 rows (tail):")
display(df.tail(2))

print("\nShape (rows, columns):", df.shape)
print("Column names:", df.columns.tolist())
print("Index (row labels):", df.index.tolist())
```

```python

print("Data types of each column:")
print(df.dtypes)

print("\nFull structural summary:")
df.info()
```


---
## 4️⃣ Selecting Columns & Rows (`loc` and `iloc`)

- **`loc`** = select by **label** (column name, index label)
- **`iloc`** = select by **position** (numeric location, like array indexing)


```python

# Selecting a single column (returns a Series)
print("Just the 'Name' column:")
print(df["Name"])
```

```python

# Selecting multiple columns
print("Name and Marks columns together:")
df[["Name", "Marks"]]
```

```python

# loc: select rows by label, and specific columns
print("Using loc — rows 0 to 2, only Name & Marks columns:")
df.loc[0:2, ["Name", "Marks"]]
```

```python

# iloc: select rows and columns purely by position (like a 2D array)
print("Using iloc — first 3 rows, first 2 columns:")
df.iloc[0:3, 0:2]
```

```python

# Slice: taking a smaller piece of the DataFrame
first_four_students = df[0:4]
print("Slice — first four students:")
first_four_students
```


---
## 5️⃣ Filtering With Conditions

- **Boolean** = a value that is `True` or `False`
- **Filter** = keeping only rows that match a condition


```python

# Boolean mask: which students scored above 80?
above_80 = df["Marks"] > 80
print("Boolean mask:")
print(above_80)
```

```python

# Filter: use the mask to keep only matching rows
top_students = df[df["Marks"] > 80]
print("Students who scored above 80:")
top_students
```

```python

# Combining multiple conditions with & (and) / | (or)
# Students from Karachi AND scored above 80
karachi_top = df[(df["City"] == "Karachi") & (df["Marks"] > 80)]
print("Students from Karachi who scored above 80:")
karachi_top
```

```python

# Query: asking a question of your data using a readable string
result = df.query("Marks > 85 and City == 'Karachi'")
print("Using .query() — same kind of question, more readable:")
result
```


---
## 6️⃣ Adding, Modifying & Dropping Columns


```python

df2 = df.copy()

# Adding a new column: Pass/Fail based on Marks
df2["Pass"] = df2["Marks"] >= 75
print("Added a new 'Pass' column:")
df2
```

```python

# Modifying an existing column: add 2 bonus marks to everyone (vectorized, no loop)
df2["Marks"] = df2["Marks"] + 2
print("Marks column after adding 2 bonus marks to everyone:")
df2[["Name", "Marks"]]
```

```python

# Dropping a column we no longer need
df2 = df2.drop(columns=["Pass"])
print("After dropping the 'Pass' column:")
df2
```

## 7️⃣ Handling Missing Data (NaN)

**NaN** = "Not a Number" — represents missing or unknown data.

> 🧠 *If a student didn't attend an exam, their score might be recorded as NaN instead of a number.*


```python

print("Which cells are missing?")
print(df.isna())

print("\nNumber of missing values per column:")
print(df.isna().sum())
```

```python

# Option 1: fill missing values (e.g. with the column mean)
df_filled = df.copy()
df_filled["Age"] = df_filled["Age"].fillna(df_filled["Age"].mean())
print("Age column after filling the missing value with the mean:")
df_filled
```

```python

# Option 2: drop rows that have any missing values
df_dropped = df.dropna()
print("After dropping rows with missing values:")
df_dropped
```


---
## 8️⃣ Descriptive Statistics

| Term | Definition |
|---|---|
| **Mean** | The average |
| **Median** | The middle value when sorted |
| **Mode** | The most frequent value |
| **Standard Deviation** | How spread out numbers are from the average |


```python

marks = df["Marks"]

print("Mean:", marks.mean())
print("Median:", marks.median())
print("Mode:", marks.mode()[0])
print("Standard Deviation:", round(marks.std(), 2))
print("Min:", marks.min(), "| Max:", marks.max())
```

```python

# describe() gives all key stats in one shot
print("Full statistical summary of numeric columns:")
df.describe()
```


---
## 9️⃣ Sorting, Duplicates & Unique Values

- **Duplicate** = a row/value that appears more than once
- **Unique** = values that appear only once, no repeats


```python

sorted_df = df.sort_values(by="Marks", ascending=False)
print("Sorted by Marks (highest first):")
sorted_df
```

```python

unique_cities = df["City"].unique()
print("Unique cities:", unique_cities)

value_counts = df["City"].value_counts()
print("\nHow many students per city:")
print(value_counts)
```

```python

print("Duplicate rows found:", df.duplicated().sum())

demo_df = pd.concat([df, df.iloc[[0]]], ignore_index=True)  # copy Ali's row again
print("After adding a duplicate row on purpose:", demo_df.duplicated().sum(), "duplicate(s) found")
```


---
## 🔟 GroupBy

**GroupBy** = splitting data into groups based on a value, so you can summarize each group.

> 🧠 *Grouping students by "City" to find the average marks in each city.*


```python

avg_marks_by_city = df.groupby("City")["Marks"].mean()
print("Average marks per city:")
avg_marks_by_city
```

```python

city_subject_summary = df.groupby(["City", "Subject"])["Marks"].mean()
print("Average marks per city, broken down by subject:")
city_subject_summary
```

```python

# Multiple statistics at once per group
summary = df.groupby("City")["Marks"].agg(["mean", "min", "max", "count"])
print("Full summary of marks per city:")
summary
```


---
## 1️⃣1️⃣ Merging, Joining & Concatenating Tables

- **Join** = combining two tables using a shared column
- **Key** = the column used to match rows between tables

> 🧠 *Merging a "Student Info" table with a "Student Attendance" table using "RollNo" they both share.*


```python

student_info = pd.DataFrame({
    "RollNo": [1, 2, 3, 4],
    "Name":   ["Ali", "Sara", "Zain", "Mina"]
})

attendance = pd.DataFrame({
    "RollNo": [1, 2, 3, 4],
    "AttendancePercent": [95, 88, 76, 99]
})

merged = pd.merge(student_info, attendance, on="RollNo")
print("Merged table (joined using 'RollNo' as the key):")
merged
```

```python

# Concatenate: stacking two tables with the SAME columns on top of each other
batch_1 = pd.DataFrame({"Name": ["Ali", "Sara"], "Marks": [85, 90]})
batch_2 = pd.DataFrame({"Name": ["Zain", "Mina"], "Marks": [76, 60]})

all_batches = pd.concat([batch_1, batch_2], ignore_index=True)
print("Concatenated (stacked) tables:")
all_batches
```


---
## 1️⃣2️⃣ Pivot Tables

**Pivot Table** = a table that summarizes and rearranges data to see patterns easily.

> 🧠 *Turning a long list of records into a summary showing average marks per city per subject.*


```python

pivot = df.pivot_table(values="Marks", index="City", columns="Subject", aggfunc="mean")
print("Pivot table — average marks by City and Subject:")
pivot
```


---
## 1️⃣3️⃣ Apply & Map — Custom Functions on Data

- **Function** = a reusable block of instructions that performs a task
- `.apply()` lets you run your own custom function on every value in a column
- `.map()` is a simpler version, often used for direct value substitutions


```python

# A custom Function: convert marks into a Grade
def marks_to_grade(marks):
    if marks >= 90:
        return "A"
    elif marks >= 75:
        return "B"
    elif marks >= 60:
        return "C"
    else:
        return "F"

df3 = df.copy()
df3["Grade"] = df3["Marks"].apply(marks_to_grade)
print("Added a 'Grade' column using a custom function with .apply():")
df3[["Name", "Marks", "Grade"]]
```

```python

# .map() for simple value substitution
city_short_codes = {"Lahore": "LHR", "Karachi": "KHI", "Islamabad": "ISB", "Multan": "MUX"}
df3["CityCode"] = df3["City"].map(city_short_codes)
print("Added a 'CityCode' column using .map():")
df3[["Name", "City", "CityCode"]]
```

## 1️⃣4️⃣ Reading & Writing CSV Files

**CSV** = "Comma-Separated Values" — a simple text file format for tables, like a basic spreadsheet.


```python

# Save our DataFrame to a CSV file
df.to_csv("students_pandas.csv", index=False)
print("Saved to students_pandas.csv ✅")

# Read it back
df_from_csv = pd.read_csv("students_pandas.csv")
print("\nData read back from CSV:")
df_from_csv
```
